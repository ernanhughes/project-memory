## The chapter question

> **What should memory keep?**

Retention is cheap; competition is not. Every retained unit competes for retrieval rank, context budget and reader attention. Growth policies — consolidate, compress, forget — decide what remains able to compete. None has yet earned its place by measurement.

## The pressure, measured

The controlled corpus behind the frame experiments is small and already budget-relevant. Scale that by years and the assembly budget becomes the binding constraint — which is why this chapter exists.


```python
from context_frames import corpus as CORPUS
from context_frames.model import estimate_tokens

units = CORPUS.corpus()
toks = sum(u.tokens for u in units)
kinds = {}
for u in units:
    kinds[u.kind] = kinds.get(u.kind, 0) + 1
print(f"Controlled corpus: {len(units)} units, ~{toks} estimated tokens")
print("Units by evidence class:")
for kind, count in sorted(kinds.items(), key=lambda kv: -kv[1]):
    print(f"  {kind:22s} {count}")
print("\nA 1500-token working budget already excludes most of this "
      "tiny corpus. Years of history make selection mandatory, and "
      "selection is what growth policies must earn.")
```

## What a result would need to show

The planning notes (superseded numbering, historical material — not current contract) sketch the shape: repeated episodes replaced by supported patterns must improve downstream behaviour **without** harmful generalisation, with exceptions, provenance and expiry preserved. Type-A: gains with preservation → consolidation earned as guarded inference. Type-B: episodes equal or generalisation unpreventable → narrow to human-approved rules or reject automatic consolidation.

Candidate operations that exist in the repository today are versioned rebuilds (derived state is disposable, raw history canonical), supersession marking (Chapter 8), echo demotion (Chapter 7) and budget-bounded assembly (Chapters 10/14). None of them is a consolidation policy. That policy is the pending experiment.


## What this establishes

- **The pressure is measured**, not asserted (corpus tokens vs working budget)
- **The target curves exist** (recall per token across budgets)
- **Candidate vocabulary exists** (rebuild, supersede, demote, assemble)
- **No growth policy is earned** — forgetting, in particular, covers distinct operations (delete evidence, hide from context, lower priority, compress, supersede) that must not be conflated


## What this does NOT establish

- Any consolidation, compression or forgetting result
- Long-term availability under years of accumulation
- Deletion/privacy propagation through derived state
- Source reliability over time
