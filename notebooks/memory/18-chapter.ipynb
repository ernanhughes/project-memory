## The earned architecture (from Chapter 18)

```mermaid
flowchart TD
    HIST[Raw project history] --> RET[Strong retrieval]
    RET --> COND[Optional derived views]
    COND --> STATE[State and evidence resolution]
    STATE --> FRAME[Current work frame]
    FRAME --> SEL[Frame-conditioned candidate retrieval]
    SEL --> CTX[Bounded context assembly]
    CTX --> READ[Reader or agent]
    READ --> BEH[Observable behaviour]
    BEH --> EVAL[Evaluation]
```

Cross-cutting channels:

```text
PROVENANCE
TEMPORAL VALIDITY
OPEN / DERIVED OBLIGATIONS
CONTEXT TRACE
POLICY VERSION
```

Fallback arrows (part of the architecture, not exceptions):

```text
derived view uncertain → raw evidence
frame uncertain → broaden toward strong RAG
current belief challenged → temporal history + provenance
maintained projection stale → recompute / reverify
policy change proposed → replay before promotion
```

## The six questions answered by the final system

Where the implementation supports them, the capstone answers:

1. **Where did we discuss X?** → Strong retrieval (Chapter 3) — CORE
2. **What did we decide about X?** → Temporal resolution + evidence lineage (Chapters 8, 7) — CORE
3. **Why did we decide X?** → Evidence lineage with support groups (Chapter 7) — CORE WHEN CLAIMS DERIVED
4. **Is X still true?** → Temporal trajectories + supersession (Chapter 8) — CORE FOR TEMPORAL
5. **What did we leave unfinished?** → Open-loop status (Chapter 9) — CONDITIONAL
6. **What from the past matters right now?** → Frame-conditioned selection + assembly (Chapters 10, 14) — CORE

## Close the loop: Measurement Instrument on capstone output

The final lesson is executable: feed the observable output into the actual measurement instrument.

````python
from experiments.benchmark.memory_measurement import MemoryTask, SystemOutput, score_task
from experiments.benchmark.memory_measurement.tasks import ANSWERABLE

# Create the task
task = MemoryTask(
    task_id="fix-store",
    family="decision",
    prompt="What should new event-store services use?",
    history_ref="ch12-fixture",
    expected_sources=("adr-007",),
    expected_state="PostgreSQL",
    superseded_options=("SQLite",),
    expected_current="PostgreSQL",
    expected_historical="SQLite",
    temporal_mode="current",
    expected_status=ANSWERABLE,
)

# Simulate the capstone C4 output (from frozen trace)
output = SystemOutput(
    task_id="fix-store",
    answer="New services should use PostgreSQL per adr-007 (July 11 decision). The team chose PostgreSQL instead, after SQLite slowed under concurrent writes.",
    retrieved_ids=("adr-007", "session-035", "session-033"),
    cited_sources=("adr-007",),
    abstained=False,
)

history_ids = ("adr-007", "session-031", "session-033", "session-035", "adr-009", "session-040", "session-044", "fact-301", "fact-302")

print("=== Measurement Instrument Scorecard (Capstone C4 Output) ===\n")
observations = score_task(task, output, history_ids)
for obs in observations:
    val = f"{obs.value:.2f}" if isinstance(obs.value, float) else obs.value
    fail = f" [{obs.failure_class}]" if obs.failure_class else ""
    print(f"  {obs.metric:30s} {val}{fail}")
    if obs.evidence:
        for e in obs.evidence:
            print(f"    → {e}")
````

## The final lesson

> **We do not call this memory because the architecture contains a vector database, graph, state store or context selector. We call it memory when retained history changes present behaviour, and the instrument lets us test whether that change helped.**

The architecture above is a candidate explanation for how the effect happens. It is not the only possible explanation, and nine controlled tasks across three readers do not settle which parts are load-bearing outside the fixture.

The investigation cut several attractive ideas that failed to justify automatic promotion:
- No single fused graph weight for truth, association, and usefulness
- No lateral inhibition (Chapter 5 ablation found no contribution)
- No learned quality router for Nexus (quality headroom was zero)
- No collapsed scalar threshold for Chapter 11's licence
- No promotion of Chapter 15/16 (experimental/deferred)

> **This is not missing ambition. It is the investigation working.**

## What remains open

The book ends without pretending the subject is closed:

- Reliable WorkFrame inference without frame capture
- Real-corpus extraction quality for graph, evidence, derived-loop structures
- Long-term availability policy under years of accumulation
- Deletion and privacy propagation through derived state
- Source reliability over time
- Multi-agent disagreement and shared memory
- Outcome attribution strong enough for safe adaptation
- Procedure extraction and precondition verification
- Continual learning across model changes
- Memory portability between readers and agents

These are not missing chapters. They are the frontier after the book's question has been made measurable.
